```python
X_train_benign = X_train[y_train == "BENIGN"]

print("Autoencoder training shape (BENIGN only):", X_train_benign.shape)
print("\nFull test set (unchanged - mixed classes for evaluation):", X_test.shape)
print(y_test.value_counts())
```

```python
scaler = StandardScaler()
X_train_benign_scaled = scaler.fit_transform(X_train_benign)
X_test_scaled = scaler.transform(X_test)

print("Re-scaled train shape:", X_train_benign_scaled.shape)
print("Re-scaled test shape:", X_test_scaled.shape)

col_stds_check = pd.Series(X_train_benign_scaled.std(axis=0), index=X_train_benign.columns)
print("Remaining near-zero variance columns:", (col_stds_check < 0.01).sum())
```

```python
# Reshape from (samples, 67) to (samples, 67, 1) for BiLSTM input
X_train_benign_reshaped = X_train_benign_scaled.reshape(
    X_train_benign_scaled.shape[0],  # number of samples (unchanged)
    X_train_benign_scaled.shape[1],  # 67 features -> becomes 67 timesteps
    1                                  # 1 value per timestep
)

X_test_reshaped = X_test_scaled.reshape(
    X_test_scaled.shape[0],
    X_test_scaled.shape[1],
    1
)

print("Train reshaped:", X_train_benign_reshaped.shape)
print("Test reshaped:", X_test_reshaped.shape)
```

```python
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Bidirectional, LSTM, RepeatVector, TimeDistributed, Dense

timesteps = X_train_benign_reshaped.shape[1]  # 67
n_features = X_train_benign_reshaped.shape[2]  # 1

# --- Encoder ---
inputs = Input(shape=(timesteps, n_features))
encoded = Bidirectional(LSTM(64, activation='relu', return_sequences=True))(inputs)
encoded = Bidirectional(LSTM(32, activation='relu', return_sequences=False))(encoded)

# --- Bottleneck to Decoder bridge ---
decoded = RepeatVector(timesteps)(encoded)

# --- Decoder ---
decoded = Bidirectional(LSTM(32, activation='relu', return_sequences=True))(decoded)
decoded = Bidirectional(LSTM(64, activation='relu', return_sequences=True))(decoded)
outputs = TimeDistributed(Dense(n_features))(decoded)

autoencoder = Model(inputs, outputs)
autoencoder.compile(optimizer='adam', loss='mse')
autoencoder.summary()
```
